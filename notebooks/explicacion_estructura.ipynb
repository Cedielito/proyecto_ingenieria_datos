# Estructura del Proyecto de Ingeniería de Datos

## 1. Descripción General

Este proyecto de Ingeniería de Datos, organizando el código y los datos en módulos separados para facilitar la escalabilidad, mantenimiento y trabajo.

## 2. Estructura del Proyecto


```
proyecto_ingenieria_datos/
│
├── data/
│   ├── raw/
│   ├── interim/
│   ├── processed/
│   └── external/
│
├── notebooks/ -> documentación y notebooks de EDA
│
├├── Project/ ← código fuente modular
│ ├── ingestion/ ← cargar/descargar datos (load, fetch)
│ ├── preprocessing/ ← limpieza y normalización
│ ├── features/ ← ingeniería de features, embeddings
│ ├── modeling/ ← entrenamiento/uso de modelos tradicionales
│ ├── llm/ ← prompts + conexión a LLM (llm_connector.py)
│ ├── pipeline/ ← orquestador (run_pipeline.py)
│ └── utils/ ← helpers, config adicional
│ ├── init.py
│ ├── config.py
│ ├── dataset.py
│ ├── features.py
│ └── plots.py
│
├── models/
│ ├── trained/ ← modelos serializados (.pkl/.pt)
│ └── embeddings/ ← vectores/embeddings
│
├── reports/
│ ├── figures/
│ ├── tables/
│ └── pdf/
│
├── tests/ ← pruebas unitarias (pytest)
├── .env ← variables sensibles (NO subir a Git)
├── .gitignore
├── Makefile / scripts
├── pyproject.toml / requirements.txt
└── README.md
```

### data/
Contiene el ciclo de vida del dato:
- **raw/** — datos originales sin tocar (CSV, JSON, PDFs). *Nunca modificar*.
- **interim/** — resultados intermedios de validaciones o transformaciones parciales.
- **processed/** — datos limpios y listos para modelado o para enviar al LLM (por ejemplo `processed.csv`).
- **external/** — datos descargados de APIs, catálogos o referencias externas.

---

### notebooks/
Notebooks de documentación y experimentación:
- EDA (análisis exploratorio)
- Pruebas rápidas
- Documentación interactiva del pipeline

---

### Project/  ← código fuente modular (lo más importante)
- **ingestion/** — funciones para cargar datos desde archivos o APIs. (ej. `load_data.py`, `api_fetch.py`)
- **preprocessing/** — limpieza y normalización (ej. `clean_text.py`, `handle_nulls.py`).
- **features/** — ingeniería de features; generación de embeddings; extracción de variables.
- **modeling/** — scripts de entrenamiento / predicción (si aplica) y serialización de modelos.
- **llm/** — construcción de prompts y conexión al LLM (`prompt_builder.py`, `llm_connector.py`). Aquí se gestionan llamados a APIs (API keys en `.env`).
- **pipeline/** — orquestador que ejecuta todo el flujo (`run_pipeline.py`).
- **utils/** — helpers, utilidades reutilizables y configuración adicional.
- Archivos raíz del paquete:
  - `__init__.py` — convierte `Project/` en paquete Python.
  - `config.py` — rutas del proyecto con `pathlib`, carga `.env`.
  - `dataset.py` — funciones de lectura/escritura (ej. `load_raw_dataframe`, `save_processed_dataframe`).
  - `features.py` — transformaciones aplicadas a los textos/datos.
  - `plots.py` — funciones para crear y guardar figuras en `reports/figures/`.

---

### models/
- **trained/** — modelos serializados (`.pkl`, `.pt`) listos para producción o evaluación.
- **embeddings/** — vectores/embeddings calculados que se reutilizan.

---

### reports/
- **figures/** — gráficas e imágenes generadas.
- **tables/** — tablas de resultados.
- **pdf/** — informes finales exportados.

---

### tests/
- Pruebas unitarias (pytest) para asegurar que las funciones críticas no rompan el pipeline.
  - Ejemplos: `test_preprocessing.py`, `test_llm.py`, `test_pipeline.py`

---

### Otros archivos importantes
- **.env** — variables sensibles (API keys). *No subir a Git*.
- **.gitignore** — ignora `.venv/`, `.env`, archivos temporales.
- **Makefile / scripts** — comandos rápidos (`make run`, `make test`).
- **pyproject.toml / uv.lock** — gestión de dependencias (si usas `uv`). Alternativa: `requirements.txt`.
- **README.md** — instrucciones para ejecutar el proyecto localmente.
